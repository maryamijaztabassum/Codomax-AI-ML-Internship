This project demonstrates a beginner-friendly machine learning workflow using **Scikit-learn**. A classification model is trained to predict the species of an Iris flower from its sepal and petal measurements.

## Objectives

- Understand the basic machine learning workflow.
- Load and explore a beginner-friendly dataset.
- Prepare features and target labels.
- Split data into training and testing sets.
- Train a classification model using Scikit-learn.
- Evaluate the model and understand prediction results.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay

print('Libraries imported successfully.')
```

## 1. Load the Dataset

```python
# If using Google Colab, upload iris_dataset.csv before running this cell.
df = pd.read_csv('iris_dataset.csv')
df.head()
```

```python
print('Dataset shape:', df.shape)
print('\nData types:')
print(df.dtypes)

print('\nMissing values:')
print(df.isnull().sum())

print('\nClass distribution:')
print(df['species'].value_counts())
```

## 2. Explore the Data

The dataset contains four numerical features: sepal length, sepal width, petal length, and petal width. The target is the Iris species.

```python
display(df.describe())

print('\nAverage measurements by species:')
display(df.groupby('species')[[
    'sepal_length_cm', 'sepal_width_cm',
    'petal_length_cm', 'petal_width_cm'
]].mean().round(2))
```

## 3. Prepare Features and Target

```python
feature_columns = [
    'sepal_length_cm', 'sepal_width_cm',
    'petal_length_cm', 'petal_width_cm'
]

X = df[feature_columns]
y = df['target']

print('Features shape:', X.shape)
print('Target shape:', y.shape)
```

## 4. Split the Dataset

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print('Training samples:', len(X_train))
print('Testing samples:', len(X_test))
```

## 5. Scale the Features

Standardization puts the numerical features on a comparable scale before model training.

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('Feature scaling completed.')
```

## 6. Train the Machine Learning Model

```python
model = LogisticRegression(max_iter=200)
model.fit(X_train_scaled, y_train)

print('Logistic Regression model trained successfully.')
```
